```python
import numpy as np
```

```python
c_pl = 6.2
P = 0.4
dstorm = 72 * 3600
xi = 0.5
tan_a = 1/1.5

def vandermeer(Hs, Tm, S):
    s_0m = (2 * np.pi * Hs) / (9.81 * Tm**2)
    xi = tan_a / np.sqrt(s_0m)
    N = dstorm / Tm
    stab_num = c_pl * P**0.18 * (S/np.sqrt(N))**0.2 * xi**-0.5
    dn50 = Hs / (stab_num)
    return dn50

def stab_num(d_50, Hs, Tm, S):
    s_0m = (2 * np.pi * Hs) / (9.81 * Tm**2)
    xi = tan_a / np.sqrt(s_0m)
    print("xi:", xi)
    N = dstorm / Tm
    stab_num = c_pl * P**0.18 * (S/np.sqrt(N))**0.2 * xi**-0.5
    return stab_num
```

```python
# ULS
S = 8
H_s = 4.1
T_m = 6.1
d_50 = 1.5

dn50_uls = vandermeer(H_s, T_m, S)

print(f"ULS: dn50 = {dn50_uls:.2f} m")
stab_num_uls = stab_num(d_50, H_s, T_m, S)
print(f"ULS: Stability number = {stab_num_uls:.2f}")
```
