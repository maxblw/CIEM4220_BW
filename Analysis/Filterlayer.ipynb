```python
import numpy as np
```

```python
def find_d50(d15, d85, C_u):
    return ((1 + C_u)/2) * d15

def find_d15(d85, C_u):
    return d85 / C_u
```

```python
# Filter 1
C_u = 1.5
C_p = 4.5 # factor for d85B to d15F conversion
d15_core = 0.17
d85_fil1 = 0.17  / C_p
d15_fil1 = d85_fil1 / C_u
d50_fil1 = find_d50(d15_fil1, d85_fil1, C_u)

print(f"Filter 1: d85 = {d85_fil1:.3f} m, d15 = {d15_fil1:.3f} m, d50 = {d50_fil1:.3f} m")

# Filter 2
d85_fil2 = d15_fil1 / C_p
d15_fil2 = d85_fil2 / C_u
d50_fil2 = find_d50(d15_fil2, d85_fil2, C_u)

print(f"Filter 2: d85 = {d85_fil2:.3f} m, d15 = {d15_fil2:.3f} m, d50 = {d50_fil2:.3f} m")

# sand check
d85_sand = 0.00075

if d15_fil2 / d85_sand < 5:
    print(f"filters OK: d15_fil2 / d85_sand = {d15_fil2 / d85_sand}")
else:
    print(f"filters NOT OK: d15_fil2 / d85_sand = {d15_fil2 / d85_sand}")



```
